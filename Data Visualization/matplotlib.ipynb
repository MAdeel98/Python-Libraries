```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
x=np.linspace(1,7,11)
x
```

```python
y= x**2
y
```

```python
plt.plot(x,y)
plt.xlabel("X-Axis")
plt.ylabel('Y-Axis')
plt.title('Papar Wala')
```

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(y,x)
plt.subplot(2,2,3)
plt.plot(y**2,x)
plt.subplot(2,2,4)
plt.plot(x**2,x**2)
```

***Object Oriented***

```python
fig=plt.figure()
axis1=fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)
axis2=fig.add_axes([0.2,0.4,.3,0.3])
axis2.plot(x,y)
```
